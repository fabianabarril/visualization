```python
import seaborn as sns
import matplotlib.pyplot as plt 
import pandas as pd
import numpy as np
import scipy
import PIL
import requests

np.random.seed(613)
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
sns.lineplot(data=tips, x="total_bill", y="tip")
```

```python
sns.lineplot(data=tips, x="total_bill", y="tip")
sns.set_style("whitegrid")
plt.show()
```

```python
tipgraph = sns.lineplot(data=tips,
x='total_bill',
y='tip')
tipgraph.set(title='Tips vs. Total Bill',
xlabel='Total Bill ($)',
ylabel='Tip Amount ($)')

sns.set_style("darkgrid")

plt.show()
```

```python
fig = plt.subplots(figsize=(10, 3))

tipgraph = sns.lineplot(data=tips,
x='total_bill',
y='tip',
color = 'hotpink',
linestyle = '--',
linewidth = 1,
marker = 'o',
markerfacecolor = 'indigo')

tipgraph.set(title='Tips vs. Total Bill',
xlabel='Total Bill ($)',
ylabel='Tip Amount ($)')

plt.show()
```

```python
tipgraph = sns.scatterplot(data=tips, x='total_bill',
y='tip', style = 'time', hue ='day', palette = ['purple', 'hotpink', 'deepskyblue', 'yellowgreen'])
tipgraph.set(title='Tips vs. Total Bill',
xlabel='Total Bill ($)',
ylabel='Tip Amount ($)')

plt.show()
```

```python
sns.pairplot(
data = tips,
hue = 'day')
```

```python
daysplot = sns.relplot(
    data=tips,        # The dataset to use (in this case, 'tips' — a built-in Seaborn dataset)
    x="total_bill",   # Column to plot on the x-axis
    y="tip",          # Column to plot on the y-axis
    hue="sex",        # Use different colors to distinguish between 'sex' categories
    palette = ['blue', 'hotpink'],
    col="day",        # Create separate plots (columns) for each 'day' value
    kind="scatter",   # Type of relational plot — here, a scatter plot
    col_wrap=4        # Wrap the plots into multiple rows after 2 columns
)
```
